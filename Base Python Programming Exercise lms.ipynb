```python
# Given list
data_list = [2, 3, 'Py', '10', 1, 'SQL', 5.5, True, 3, 'John', None, 7]

# Initialize a dictionary to hold the counts of each datatype
type_count = {}

# Loop through each element in the list
for item in data_list:
    # Get the type of the item
    item_type = type(item)
    
    # Increment the count for this type in the dictionary
    if item_type in type_count:
        type_count[item_type] += 1
    else:
        type_count[item_type] = 1

# Display the counts of each datatype
for data_type, count in type_count.items():
    print(f"{data_type.__name__}: {count}")
```

```python
def descriptive_statistics(numbers):
    # Check if the input list is empty
    if not numbers:
        return "The list is empty."

    # Calculate the required statistics
    total_sum = sum(numbers)
    count = len(numbers)
    minimum = min(numbers)
    maximum = max(numbers)
    mean = total_sum / count

    # Return the results as a dictionary
    return {
        'sum': total_sum,
        'count': count,
        'min': minimum,
        'mean': mean,
        'max': maximum
    }

# Example usage
input_numbers = [10, 20, 30, 40, 50]
result = descriptive_statistics(input_numbers)
print(result)
```

```python
def calculate_area(shape, **kwargs):
    if shape == 'square':
        side = kwargs.get('side')
        if side is not None:
            return side ** 2
        else:
            return "Please enter 'side' for a square."
    
    elif shape == 'rectangle':
        length = kwargs.get('length')
        width = kwargs.get('width')
        if length is not None and width is not None:
            return length * width
        else:
            return "Please enter 'length' and 'width' for a rectangle."
    
    elif shape == 'circle':
        radius = kwargs.get('radius')
        if radius is not None:
            return 3.14159 * (radius ** 2)  # Using π ≈ 3.14159
        else:
            return "Please enter 'radius' for a circle."
    
    else:
        return "Unsupported shape. Please use 'square', 'rectangle', or 'circle'."

# Example usage
print(calculate_area('square', side=4))          # Output: 16
print(calculate_area('rectangle', length=5, width=3))  # Output: 15
print(calculate_area('circle', radius=3))        # Output: 28.27431
print(calculate_area('square'))                    # Output: Please enter 'side' for a square.
print(calculate_area('triangle', base=5, height=3))  # Output: Unsupported shape. Please use 'square', 'rectangle', or 'circle'.
```

```python
def is_prime(n):
    # Check if the number is less than 2
    if n <= 1:
        return False
    # Check for factors from 2 to the square root of n
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

# Example usage
number = 29
if is_prime(number):
    print(f"{number} is a prime number.")
else:
    print(f"{number} is not a prime number.")
```

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)

# Get user input
number = int(input("Enter a number: "))

# Calculate factorial
result = factorial(number)

# Print the result
print(result)
```

```python
# Get user input
input_words = input("Enter a sequence of comma-separated words: ")

# Generate a list from the input
word_list = input_words.split(',')

# Sort the list alphabetically
word_list.sort()

# Join the sorted list into a comma-separated string
sorted_words = ','.join(word_list)

# Print the result
print(sorted_words)
```

```python
# Get user input
input_words = input("Enter a sequence of whitespace-separated words: ")

# Generate a list from the input
word_list = input_words.split()

# Remove duplicates by converting the list to a set
unique_words = set(word_list)

# Sort the unique words alphanumerically
sorted_words = sorted(unique_words)

# Join the sorted list into a whitespace-separated string
result = ' '.join(sorted_words)

# Print the result
print(result)
```

```python
# Get user input
input_sentence = input("Enter a sentence: ")

# Initialize counters
upper_case_count = 0
lower_case_count = 0

# Iterate through each character in the input sentence
for char in input_sentence:
    if char.isupper():
        upper_case_count += 1
    elif char.islower():
        lower_case_count += 1

# Print the results
print(f"UPPER CASE {upper_case_count} LOWER CASE {lower_case_count}")
```
